```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np
cars_df=pd.read_csv('/content/Cars93 (10).csv')
print(cars_df)
```

```python
ax=cars_df.plot(['Weight'],['MPG.city'],kind="scatter",color="black")
ax.set_xlabel("weight")
ax.set_ylabel("MPG.city")
ax.set_title("weightvsmpg.city",fontsize=16)
```

```python
plt.plot(cars_df['EngineSize'],cars_df['Horsepower'])
plt.xlabel('Enginesize')
plt.ylabel('Horsepower')
plt.title('line chart to observe the variations in Enginesize again Horsepower')
plt.show()
```
