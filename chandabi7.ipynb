```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
cars_df=pd.read_csv('mtcars (2).csv')
print(cars_df)
```

```python
plt.scatter (cars_df['mpg'],cars_df['wt'],marker='*')
print()
plt.xlabel('mpg')
plt.ylabel('weight')
plt.title('relationship between the weight of the cars and mpg')
plt.show()
```

```python
trans=cars_df['am'].value_counts()
print(trans)
trans.plot.bar()
plt.xlabel("transmission")
plt.title('Bar plot to check the Frequency distrtribution of Transmission type')
plt.show()
```

```python
cars_df['mpg'].plot(kind='box')
plt.xlabel('MPG')
plt.title('Box and whisker plot of MPG')
plt.show()
```

```python
minvalue=np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
maxvalue=np.max(cars_df['mpg'])
print('minimum value:',minvalue)
print('Q1:','Q1')
print('Median:',median)
print('Q3:','q3')
print('maximum value:',maxvalue)
```
